```python
x = df.loc[df['origin'] == 'Asia', ['vehicle_weight', 'model_year']].head(7).values
```

```python
x
```

```python
xt = np.transpose(x)
xt
```

```python
xtx = xt @ x 
xtx
```

```python
ixtx = np.linalg.inv(xtx)
ixtx
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
y
```

```python
w = (ixtx @ xt) @ y
w
```
